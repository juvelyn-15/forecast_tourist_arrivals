# Segment Analysis

Segments are treated as supporting analysis for market composition and heterogeneous recovery. They are not primary forecasting targets in the main model comparison.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / "src").exists() and (ROOT.parent / "src").exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from src.config import CLEAN_SEGMENTS, FIGURES, TABLES, MAIN_TARGET, SEGMENT_COLUMNS, SERIES_COLORS, SERIES_LABELS
from src.plotting import annotate_events, save_figure, set_academic_style
set_academic_style()
df = pd.read_csv(CLEAN_SEGMENTS, parse_dates=["date"]).set_index("date").sort_index()
df.index.freq = "MS"
```

## Segment Volume Structure

```python
fig, ax = plt.subplots()
for col in SEGMENT_COLUMNS:
    ax.plot(df.index, df[col], color=SERIES_COLORS[col], label=SERIES_LABELS[col])
ax.set_title("Segment Arrival Volumes as Supporting Market Evidence")
ax.set_ylabel("Monthly arrivals")
ax.legend(ncol=3, fontsize=8)
plt.show()
```

Observation -> segment volumes differ substantially in level and recovery path. Statistical implication -> aggregate behavior masks source-market heterogeneity. Tourism implication -> strategic planning should distinguish source-market structure from aggregate arrival forecasting.

## Segment Shares

```python
shares = df[SEGMENT_COLUMNS].div(df[MAIN_TARGET], axis=0) * 100
fig, ax = plt.subplots()
for col in SEGMENT_COLUMNS:
    ax.plot(shares.index, shares[col], color=SERIES_COLORS[col], label=SERIES_LABELS[col])
ax.set_title("Segment Shares in Total International Arrivals")
ax.set_ylabel("Share (%)")
ax.legend(ncol=3, fontsize=8)
plt.show()
```

Observation -> shares vary over time and across shock regimes. Statistical implication -> composition shifts are analytically relevant even when segments are not forecast targets. Tourism implication -> recovery policy should be source-market-specific.
